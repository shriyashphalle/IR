```python
# Step 1: Import required libraries and download necessary resources
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
# Step 2: Define sample text and perform word tokenization
sample_text = """
Natural Language Processing is an exciting field.
The developers are coding, testing, and developing amazing applications daily.
He likes to run, and he ran yesterday during his running session.
"""

# Tokenize and convert to lowercase
all_tokens = word_tokenize(sample_text.lower())
print("Tokenized Words:", all_tokens)
```

```python
# Step 3: Filter out stop words and non-alphabetic punctuation
english_stopwords = set(stopwords.words('english'))

clean_tokens = [token for token in all_tokens if token.isalnum() and token not in english_stopwords]
print("After Stop Word Removal:", clean_tokens)
```

```python
# Step 4: Apply stemming to reduce words to their root forms
porter_stemmer = PorterStemmer()

processed_stems = [porter_stemmer.stem(token) for token in clean_tokens]
print("After Stemming:", processed_stems)
```
